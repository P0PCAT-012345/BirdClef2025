```python
import os
import h5py
import pandas as pd
from tqdm import tqdm
import numpy as np
import matplotlib.pyplot as plt
```

```python
from helper_functions import *
```

```python
original_df = pd.read_csv('processed_data/expanded_df_formatted.csv')
original_df = original_df.drop(columns=['secondary_labels', 'matches_primary', 'matches_secondary', 'top_class'])

spec_averages = np.load('processed_data/spectrogram_averages.npy')
h5_path = 'processed_data/expanded_spectrograms.h5'
```

```python
original_df['spec_averages'] = spec_averages
original_df
```

```python
first_n = 5
num_classes = original_df['primary_label'].nunique()

top_n_samples = original_df.sort_values('spec_averages', ascending=False).groupby('primary_label').head(first_n)
assert len(top_n_samples) <= first_n * num_classes

top_n_samples
```

```python
save_filtered_data(original_df=original_df,
                   filtered_df=top_n_samples,
                   name=f'top_{first_n}_samples',
                   save_npy=True)
```

```python
best_n = 1 # matches the number of training samples in the baseline
best_samples_per_file = original_df.sort_values('spec_averages').groupby('filepath').head(best_n)
best_samples_per_file
```

```python
save_filtered_data(original_df=original_df,
                   filtered_df=best_samples_per_file,
                   name='best_samples_per_file',
                   save_npy=True)
```
