```python
def audio2melspec(audio_data):
    if np.isnan(audio_data).any():
        mean_signal = np.nanmean(audio_data)
        audio_data = np.nan_to_num(audio_data, nan=mean_signal)

    mel_spec = librosa.feature.melspectrogram(
        y=audio_data,
        sr=config.FS,
        n_fft=config.N_FFT,
        hop_length=config.HOP_LENGTH,
        n_mels=config.N_MELS,
        fmin=config.FMIN,
        fmax=config.FMAX,
        power=2.0
    )

    mel_spec_db = librosa.power_to_db(mel_spec, ref=np.max)
    mel_spec_norm = (mel_spec_db - mel_spec_db.min()) / (mel_spec_db.max() - mel_spec_db.min() + 1e-8)
    
    return mel_spec_norm
```

```python
# 2. Separate to 10 second chunks then use cyclic padding if the last chunk isn't 10 seconds long

def process_audio_to_10s_chunks(filepath, target_duration=10.0, sr=32000):
    """Load audio and split into exact 10-second chunks with CYCLIC PADDING if needed"""
    audio_data, _ = librosa.load(filepath, sr=sr)
    target_samples = int(target_duration * sr)
    
    if len(audio_data) < target_samples: # files shorter than target duration (safety measure)
        repeats = int(np.ceil(target_samples / len(audio_data)))
        padded_audio = np.tile(audio_data, repeats)[:target_samples]
        return [padded_audio]
    
    num_chunks = int(np.ceil(len(audio_data) / target_samples)) # longer files
    chunks = []
    for i in range(num_chunks):
        start = i * target_samples
        end = start + target_samples
        chunk = audio_data[start:end]
        
        if len(chunk) < target_samples: # cyclic padding for last chunk
            remaining = target_samples - len(chunk)
            chunk = np.concatenate([
                chunk,
                audio_data[:remaining]
            ])
        
        chunks.append(chunk)
    
    return chunks
```

## Creating expanded dataset

```python
import h5py
```

```python
expanded_dataset_filename = os.path.join(config.OUTPUT_DIR, 'expanded_spectrograms.h5')
with h5py.File(expanded_dataset_filename, 'w') as hf:
    max_shape = (len(expanded_df), *config.TARGET_SHAPE)

    chunk_size = 1000
    chunk_shape = (chunk_size, *config.TARGET_SHAPE)

    dset = hf.create_dataset(
        'spectrograms',
        shape=(0, *config.TARGET_SHAPE),
        maxshape=max_shape,
        chunks=chunk_shape,
        dtype='float32',
        compression='gzip'
    )

    batch_size = chunk_size

    prev_working_idx = -1
    prev_audio_chunks = []
    
    for i in tqdm(range(0, len(expanded_df), batch_size)):
        batch_chunks = expanded_df.iloc[i:i+batch_size]
        batch_specs = []

        for _, row in batch_chunks.iterrows():
            if row.working_df_idx == prev_working_idx:
                audio_chunk = prev_audio_chunks[row.chunk_id]

            else:
                filepath = row.filepath.replace('/kaggle/input/birdclef-2025/', config.DATA_ROOT)
                audio_chunks = process_audio_to_10s_chunks(filepath, target_duration=10.0, sr=config.FS)

                assert row.chunk_id == 0, "Make sure you're refreshing chunks only at the beginning."
                audio_chunk = audio_chunks[row.chunk_id]

                prev_working_idx = row.working_df_idx
                prev_audio_chunks = audio_chunks

            mel_spec = audio2melspec(audio_chunk)
                
            if mel_spec.shape != config.TARGET_SHAPE:
                mel_spec = cv2.resize(mel_spec, config.TARGET_SHAPE, interpolation=cv2.INTER_LINEAR)

            batch_specs.append(mel_spec)
        
        dset.resize(dset.shape[0] + len(batch_specs), axis=0)
        dset[-len(batch_specs):] = np.stack(batch_specs)
    
    # hf.create_dataset('metadata', data=expanded_df.to_records(index=False))
```

```python
all_bird_data = {}
errors = []

for _, row in tqdm(strict_labels.iterrows(), total=len(strict_labels)):
    try:
        chunks = process_audio_to_10s_chunks(row.filepath, target_duration=10.0, sr=config.FS)
        assert len(chunks) == row.chunk_id_count, f"Expected {row.chunk_id_count}, Found {len(chunks)}, Target: {row.chunk_id}"
        
        chunk = chunks[row.chunk_id]

        mel_spec = audio2melspec(chunk)
        if mel_spec.shape != config.TARGET_SHAPE:
            mel_spec = cv2.resize(mel_spec, config.TARGET_SHAPE, interpolation=cv2.INTER_LINEAR)

        all_bird_data[row.original_samplename] = mel_spec.astype(np.float32)

    except Exception as e:
        print(f"Error processing {row.filepath}: {e}")
        errors.append((row.filepath, str(e)))
```

```python
print("Starting audio processing...")
print(f"{'DEBUG MODE - Processing only 50 samples' if config.DEBUG_MODE else 'FULL MODE - Processing all samples'}")
start_time = time.time()

all_bird_data = {}
errors = []

for i, row in tqdm(expanded_df.iterrows(), total=total_samples):
    if config.N_MAX is not None and i >= config.N_MAX:
        break
    
    try:
        audio_data, _ = librosa.load(row.filepath, sr=config.FS)
        primary_label = row.primary_label
        
        secondary_labels = [e.replace("'", '') for e in row.secondary_labels]
        secondary_labels = None if secondary_labels[0] == '' else secondary_labels

        target_samples = int(config.TARGET_DURATION * config.FS)

        if len(audio_data) < target_samples:
            n_copy = math.ceil(target_samples / len(audio_data))
            if n_copy > 1:
                audio_data = np.concatenate([audio_data] * n_copy)

        start_idx = max(0, int(len(audio_data) / 2 - target_samples / 2))
        end_idx = min(len(audio_data), start_idx + target_samples)
        center_audio = audio_data[start_idx:end_idx]

        if len(center_audio) < target_samples:
            center_audio = np.pad(center_audio, 
                                 (0, target_samples - len(center_audio)), 
                                 mode='constant')

        mel_spec = audio2melspec(center_audio)

        if mel_spec.shape != config.TARGET_SHAPE:
            mel_spec = cv2.resize(mel_spec, config.TARGET_SHAPE, interpolation=cv2.INTER_LINEAR)

        all_bird_data[row.samplename] = mel_spec.astype(np.float32)
        
    except Exception as e:
        print(f"Error processing {row.filepath}: {e}")
        errors.append((row.filepath, str(e)))

end_time = time.time()
print(f"Processing completed in {end_time - start_time:.2f} seconds")
print(f"Successfully processed {len(all_bird_data)} files out of {total_samples} total")
print(f"Failed to process {len(errors)} files")
```
